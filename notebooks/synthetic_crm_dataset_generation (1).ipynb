```python
industries = [
    "SaaS",
    "FinTech",
    "Healthcare",
    "EdTech",
    "E-commerce",
    "Manufacturing",
    "Real Estate",
    "Marketing",
    "Logistics",
    "HR Tech"
]

company_sizes = [
    "Small",
    "Medium",
    "Enterprise"
]

countries = [
    "India",
    "USA",
    "UK",
    "Singapore",
    "Australia"
]
```

```python
lead_data = []

for lead_id in range(1, 10001):

    campaign_id = random.choices(
        population=[1,2,3,4,5],
        weights=[30,25,15,20,10],
        k=1
    )[0]

    lead_source = {
        1:"LinkedIn",
        2:"Google Ads",
        3:"Referral",
        4:"Cold Email",
        5:"Organic Website"
    }[campaign_id]

    company_size = random.choices(
        ["Small","Medium","Enterprise"],
        weights=[55,30,15],
        k=1
    )[0]

    lead_data.append({
        "lead_id": lead_id,
        "company_name": fake.company(),
        "industry": random.choice(industries),
        "company_size": company_size,
        "country": random.choice(countries),
        "lead_source": lead_source,
        "campaign_id": campaign_id,
        "assigned_sales_rep_id": random.randint(1,10),
        "created_date": fake.date_between(
            start_date="-2y",
            end_date="today"
        )
    })
```

```python
leads = pd.DataFrame(lead_data)

leads.head()
```

```python
leads.to_csv("leads.csv", index=False)
```

```python
activity_types = [
    "Call",
    "Email",
    "LinkedIn Message",
    "WhatsApp",
    "Demo Meeting"
]

activity_outcomes = [
    "Interested",
    "Follow-up Needed",
    "No Response",
    "Demo Scheduled",
    "Rejected"
]
```

```python
activities = []

activity_id = 1

for _, lead in leads.iterrows():

    num_activities = random.randint(1,6)

    last_date = lead["created_date"]

    for i in range(num_activities):

        activity_date = last_date + timedelta(
            days=random.randint(1,10)
        )

        last_date = activity_date

        activities.append({
            "activity_id": activity_id,
            "lead_id": lead["lead_id"],
            "sales_rep_id": lead["assigned_sales_rep_id"],
            "activity_date": activity_date,
            "activity_type": random.choice(activity_types),
            "response_time_hours": random.randint(1,72),
            "outcome": random.choice(activity_outcomes)
        })

        activity_id += 1
```

```python
activities = pd.DataFrame(activities)

activities.head()
```

```python
conversion_rates = {
    "Referral": 0.40,
    "LinkedIn": 0.30,
    "Organic Website": 0.22,
    "Google Ads": 0.18,
    "Cold Email": 0.10
}

deals = []

deal_id = 1

for _, lead in leads.iterrows():

    probability = conversion_rates[lead["lead_source"]]
    random_number = random.random()

    # Decide current stage
    if random_number < probability:
        stage = "Won"
        status = "Closed Won"

    elif random_number < probability + 0.10:
        stage = "Negotiation"
        status = "Open"

    elif random_number < probability + 0.20:
        stage = "Proposal Sent"
        status = "Open"

    elif random_number < probability + 0.35:
        stage = "Demo Scheduled"
        status = "Open"

    elif random_number < probability + 0.55:
        stage = "Qualified"
        status = "Open"

    elif random_number < probability + 0.80:
        stage = "Contacted"
        status = "Open"

    else:
        stage = "Lost"
        status = "Closed Lost"

    # Product assignment
    if lead["company_size"] == "Enterprise":
        product_id = 3
        deal_value = random.randint(15000, 50000)

    elif lead["company_size"] == "Medium":
        product_id = random.choice([2, 3])
        deal_value = random.randint(5000, 18000)

    else:
        product_id = random.choice([1, 2])
        deal_value = random.randint(1000, 7000)

    deals.append({
        "deal_id": deal_id,
        "lead_id": lead["lead_id"],
        "sales_rep_id": lead["assigned_sales_rep_id"],
        "product_id": product_id,
        "stage": stage,
        "status": status,
        "deal_value": deal_value if stage == "Won" else 0,
        "close_date": lead["created_date"] + timedelta(days=random.randint(7,60))
    })

    deal_id += 1
```

```python
deals = pd.DataFrame(deals)

deals.head()
```
